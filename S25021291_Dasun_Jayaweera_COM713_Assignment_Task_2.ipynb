```python
#Remove HTML Tags

def remove_html(text):
    return re.sub(r'<.*?>', ' ', text)

df['email_text'] = df['email_text'].apply(remove_html)
```

```python
#verify

df['email_text'].head()
```

```python
#Remove Punctuation and Numbers

def remove_punctuation_numbers(text):
    return re.sub(r'[^a-zA-Z\s]', ' ', text)

df['email_text'] = df['email_text'].apply(remove_punctuation_numbers)
```

```python
#Remove Extra Spaces

def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()

df['email_text'] = df['email_text'].apply(remove_extra_spaces)
```

```python
#NLTK - install and download the stop-word list

import nltk

nltk.download('stopwords')
```

```python
#Remove Stop Words

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)

df['email_text'] = df['email_text'].apply(remove_stopwords)
```

```python
#Lemmatization

nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    words = text.split()
    words = [lemmatizer.lemmatize(word) for word in words]
    return " ".join(words)

df['email_text'] = df['email_text'].apply(lemmatize_text)
```

```python
#shape of the TF-IDF matrix

from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1,2)
)

X_text = vectorizer.fit_transform(df['email_text'])

print(X_text.shape)
```

```python
#features (words and phrases)

feature_names = vectorizer.get_feature_names_out()

print(feature_names[:30])
```

```python
#Create the Phishing Dictionary

phishing_weights = {
    "verify": 2.0,
    "password": 2.5,
    "bank": 2.2,
    "account": 1.8,
    "login": 2.0,
    "security": 1.8,
    "urgent": 2.2,
    "click": 1.5,
    "update": 1.5,
    "confirm": 1.8
}
```

```python
#Build the Weighting Function

def calculate_email_weight(text, url_count):

    score = 1.0

    words = text.split()

    for word in words:

        if word in phishing_weights:

            score *= phishing_weights[word]

    # Context Rule
    if url_count > 0:

        score *= 1.2

    return score
```

```python
#Apply to Every Email

df['context_weight'] = df.apply(
    lambda row: calculate_email_weight(
        row['email_text'],
        row['urls']
    ),
    axis=1
)
```

```python
#Convert context_weight into a Matrix

from scipy.sparse import csr_matrix

context_feature = csr_matrix(df[['context_weight']].values)
```

```python
#Combine the Features

from scipy.sparse import hstack

X = hstack([X_text, context_feature])
```

```python
#Check the Shape

print(X.shape)
```

```python
#Prepare the Target Variable

y = df['label']
```

```python
#Split the Dataset

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
#Verify

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

```python
#Train the Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier
```

```python
#Create the Model

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

```python
#Train the Model

rf_model.fit(X_train, y_train)
```

```python
#Make Predictions

y_pred = rf_model.predict(X_test)
```

```python
#Evaluate the Model

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
#Accuracy

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
```

```python
#Precision

precision = precision_score(y_test, y_pred)

print("Precision :", precision)
```

```python
#Recall

recall = recall_score(y_test, y_pred)

print("Recall :", recall)
```

```python
#F1-Score

f1 = f1_score(y_test, y_pred)

print("F1 Score :", f1)
```

```python
#Classification Report

print(classification_report(y_test, y_pred))
```

```python
#Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
#Visualise the Confusion Matrix

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

disp = ConfusionMatrixDisplay(confusion_matrix=cm)

disp.plot(cmap='Blues')

plt.title("Confusion Matrix - Random Forest")
plt.show()
```
